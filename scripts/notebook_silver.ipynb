```python

caixin_services = spark.sql(
  """
SELECT

  to_date(
    regexp_replace(date, '\\s*\\(.*\\)$', ''), 
    'dd.MM.yyyy'
  ) AS date,
  cast(regexp_replace(actual_state, ',', '.') AS DECIMAL(10,2))   AS actual_state,
  cast(regexp_replace(close, ',', '.')        AS DECIMAL(10,2))   AS close,
  cast(regexp_replace(forecast, ',', '.')     AS DECIMAL(10,2))   AS forecast
from dtb_thera.thera_raw.caixin_services
  """)
caixin_services.write.mode('overwrite').saveAsTable('dtb_thera.thera_silver.caixin_services')
```

```python



currencies = spark.sql(
  """
    select 
    try_to_timestamp(date, 'yyyyMMdd') as data_refe,
    cast(close as decimal(10,2)) as close,
    cast(open as decimal(10,2)) as open,
    cast(high as decimal(10,2)) as high,
    cast(low as decimal(10,2)) as low
    from dtb_thera.thera_raw.currencies

  """)   
currencies.write.mode('overwrite').saveAsTable('dtb_thera.thera_silver.currencies')
```

```python
historico = spark.sql(
  """
    select 
        try_to_timestamp(date, 'yyyyMMdd') as data_refe,
        cast(close as decimal(10,2)) as close,
        cast(open as decimal(10,2)) as open,
        cast(high as decimal(10,2)) as high,
        cast(low as decimal(10,2)) as low
    from dtb_thera.thera_raw.historico

  """)
historico.write.mode('overwrite').saveAsTable('dtb_thera.thera_silver.historico')
```
